# QC per sample, before pooling

```python
from pathlib import Path
import sys, json
import pandas as pd
from IPython.display import display, Image
ROOT = Path.cwd()
if not (ROOT / 'run_analysis.py').exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
def table(relative, **kwargs):
    return pd.read_csv(ROOT / relative, **kwargs)
def figure(relative):
    display(Image(filename=str(ROOT / 'figures' / relative)))
```

The filters are prespecified exploratory thresholds: 200–5,000 genes and ≤15% mitochondrial counts. The pre-filter distributions are inspected below. Do not rewrite this as retrospective threshold optimization.

```python
qc = table('results/qc_summary.csv')
display(qc)
print('Input:', qc.input_cells.sum(), 'Retained:', qc.retained_cells.sum())
```

```python
figure('qc/per_sample.png')
figure('qc/counts_vs_genes.png')
```

```python
sensitivity = table('results/qc_threshold_sensitivity.csv')
display(sensitivity.pivot(index='sample_id', columns=['max_genes', 'max_mt'], values='retained_cells'))
```

The alternative thresholds evaluate retention, not full reclustering stability. High gene count filtering does not establish doublet removal.
