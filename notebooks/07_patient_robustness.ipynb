# Held-out samples, uncertainty, and exploratory trajectory

```python
from pathlib import Path
import sys, json
import pandas as pd
from IPython.display import display, Image
ROOT = Path.cwd()
if not (ROOT / 'run_analysis.py').exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
def table(relative, **kwargs):
    return pd.read_csv(ROOT / relative, **kwargs)
def figure(relative):
    display(Image(filename=str(ROOT / 'figures' / relative)))
```

Each cancer sample is withheld while a 20-gene panel is selected using the other two cancer samples and all noncancer samples. Eligibility and selection use training samples only. Upstream annotation used the full cohort; this is conditional panel transfer, not fully nested validation or an independent test cohort.

```python
folds = table('results/signatures/lopo_cancer.csv')
display(folds)
figure('progression/lopo_cancer.png')
```

```python
selected = table('results/signatures/lopo_selected_genes.csv')
display(selected.groupby('gene').heldout_sample.nunique().sort_values(ascending=False).head(20))
```

A positive held-out contrast suggests transfer within this small cohort. A failed fold is evidence of heterogeneity, not a reason to discard that sample. The unit test changes held-out expression drastically and verifies that the selected training panel is unchanged.

```python
display(table('results/signatures/lopo_selected_genes.csv').groupby('heldout_sample').gene.apply(list))
```
