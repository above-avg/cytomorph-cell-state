```python
from pathlib import Path
import sys, json
import pandas as pd
from IPython.display import display, Image
ROOT = Path.cwd()
if not (ROOT / 'run_analysis.py').exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
def table(relative, **kwargs):
    return pd.read_csv(ROOT / relative, **kwargs)
def figure(relative):
    display(Image(filename=str(ROOT / 'figures' / relative)))
```

Epithelial cells are selected from provisional broad labels, then given new HVGs, PCA, graphs, UMAP and Leiden clusters. Cluster numbers are not stage names.
